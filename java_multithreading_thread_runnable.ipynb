# Java Multithreading: Thread Class and Runnable Interface

## AIM
To create multiple threads in Java using both methods—extending the `Thread` class and implementing the `Runnable` interface. The threads print messages at intervals of 1, 2, and 3 seconds to demonstrate concurrent execution.

## Problem Statement
Create three threads:
- **Good Morning Thread:** prints `Good Morning` every 1 second.
- **Hello Thread:** prints `Hello` every 2 seconds.
- **Welcome Thread:** prints `Welcome` every 3 seconds.

Implement the program using:
1. Extending the `Thread` class.
2. Implementing the `Runnable` interface.

## Input
No user input. The program runs automatically.

## Expected Output
Messages appear concurrently at different intervals. Their exact order may vary, for example:
```text
Good Morning
Hello
Welcome
Good Morning
Good Morning
Hello
...
```

## Algorithm
### Method 1 — Extending `Thread`
1. Create three classes that extend `Thread`.
2. Override `run()` in each class.
3. Print the required message and call `Thread.sleep()` for the required interval inside a loop.
4. Create the thread objects in `main()`.
5. Call `start()` on each object.

### Method 2 — Implementing `Runnable`
1. Create three classes that implement `Runnable`.
2. Override `run()` with the same message and timing logic.
3. Create a `Thread` object for each runnable.
4. Call `start()` on each thread object.

**Note:** These examples run indefinitely. Stop execution manually when needed. In the code below, interruption is handled by restoring the interrupt flag and exiting the loop.


## Method 1: Extending the Thread Class

````java
class GoodMorningThread extends Thread {
    @Override
    public void run() {
        while (!Thread.currentThread().isInterrupted()) {
            System.out.println("Good Morning");
            try {
                Thread.sleep(1000);
            } catch (InterruptedException e) {
                Thread.currentThread().interrupt();
                break;
            }
        }
    }
}

class HelloThread extends Thread {
    @Override
    public void run() {
        while (!Thread.currentThread().isInterrupted()) {
            System.out.println("Hello");
            try {
                Thread.sleep(2000);
            } catch (InterruptedException e) {
                Thread.currentThread().interrupt();
                break;
            }
        }
    }
}

class WelcomeThread extends Thread {
    @Override
    public void run() {
        while (!Thread.currentThread().isInterrupted()) {
            System.out.println("Welcome");
            try {
                Thread.sleep(3000);
            } catch (InterruptedException e) {
                Thread.currentThread().interrupt();
                break;
            }
        }
    }
}

public class ThreadClassDemo {
    public static void main(String[] args) {
        new GoodMorningThread().start();
        new HelloThread().start();
        new WelcomeThread().start();
    }
}
````

### Output (sample)
The messages print repeatedly. Because the threads execute concurrently, the exact order can differ on each run.

```text
Good Morning
Hello
Welcome
Good Morning
Good Morning
Hello
Welcome
...
```


## Method 2: Implementing the Runnable Interface

````java
class GoodMorningRunnable implements Runnable {
    @Override
    public void run() {
        while (!Thread.currentThread().isInterrupted()) {
            System.out.println("Good Morning");
            try {
                Thread.sleep(1000);
            } catch (InterruptedException e) {
                Thread.currentThread().interrupt();
                break;
            }
        }
    }
}

class HelloRunnable implements Runnable {
    @Override
    public void run() {
        while (!Thread.currentThread().isInterrupted()) {
            System.out.println("Hello");
            try {
                Thread.sleep(2000);
            } catch (InterruptedException e) {
                Thread.currentThread().interrupt();
                break;
            }
        }
    }
}

class WelcomeRunnable implements Runnable {
    @Override
    public void run() {
        while (!Thread.currentThread().isInterrupted()) {
            System.out.println("Welcome");
            try {
                Thread.sleep(3000);
            } catch (InterruptedException e) {
                Thread.currentThread().interrupt();
                break;
            }
        }
    }
}

public class RunnableDemo {
    public static void main(String[] args) {
        new Thread(new GoodMorningRunnable()).start();
        new Thread(new HelloRunnable()).start();
        new Thread(new WelcomeRunnable()).start();
    }
}
````

### Output (sample)
```text
Good Morning
Hello
Welcome
Good Morning
Good Morning
Hello
...
```

## Result
Thus, multiple threads were created and executed concurrently using both the `Thread` class and the `Runnable` interface, with messages printed at intervals of 1, 2, and 3 seconds.
